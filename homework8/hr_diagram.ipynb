```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
#3.2
data = np.loadtxt("/Users/lilly/Python DeCal/homework_master/course_assignments/homework7/star_data.txt", skiprows=1)
print(data.shape)
print("there are 2642 rows")

magnitude = data[:, 0]
temperature = data[:,1]

print("minimum magnitude:", magnitude.min())
print("maximum magnitude:", magnitude.max())
print("the mimimum value corresponds to the brightest star, while the maximum value corresponds to the dimmest star")

print("minimum temperature:", temperature.min())
print("maximum temperature:", temperature.max())
print("the maximum value corresponds to the hottest star, while the minimum value corresponds to the coldest star. ")

brightest_star = np.argmin(magnitude)
dimmest_star = np.argmax(magnitude)

print(f"the brightest star is star number {brightest_star}")
print(f"the dimmest star is star number {dimmest_star}")
```

```python
#3.3
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

brightest_idx = np.argmin(magnitude)
dimmest_idx = np.argmax(magnitude)
hottest_idx = np.argmax(temperature)
coldest_idx = np.argmin(temperature)

plt.figure(figsize=(8, 6))
plt.scatter(
    log_temperature,
    log_magnitude,
    c=log_temperature,
    cmap='coolwarm',
    label='stars',
    alpha=0.8
)

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.scatter(log_temperature[brightest_idx], log_magnitude[brightest_idx], 
            color='black', marker='*', s=300, label='Brightest Star')

plt.scatter(log_temperature[dimmest_idx], log_magnitude[dimmest_idx], 
            color='purple', marker='*', s=300, label='Dimmest Star')

plt.scatter(log_temperature[hottest_idx], log_magnitude[hottest_idx], 
            color='green', marker='*', s=300, label='Hottest Star')

plt.scatter(log_temperature[coldest_idx], log_magnitude[coldest_idx], 
            color='orange', marker='*', s=300, label='Coldest Star')

plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.show
```

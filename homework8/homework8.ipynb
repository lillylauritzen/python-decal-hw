```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
def sine_wave(x, A, w):
    return A * np.sin(w * x)

x = np.linspace(0, 2 * np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
fig, axes = plt.subplots(1, 3, figsize=(25,5))

for A, w in zip(amplitude, frequency):
    line_style = ['-', '--', '-.', ':', '-'][frequency.index(w)]
    y1 = sine_wave(x, A, 1)
    axes[0].plot(
        x, y1,
        linestyle=line_style,
        label=f'Amplitude = {A}, Frequency =1'
    )

    y2 = sine_wave(x, 1, w)
    axes[1].plot(
        x, y2,
        linestyle=line_style,
        label=f'Amplitude = 1, Frequency = {w}'
    )

    y3 = sine_wave(x, A, w)
    axes[2].plot(
        x, y3,
        linestyle=line_style, 
        label=f'Amplitude = {A}, Frequency = {w}'
    )

axes[0].set_title('Varying Amplitude', fontsize=16)
axes[0].set_xlabel('X Values', fontsize=14)
axes[0].set_ylabel('Y Values', fontsize=14)
axes[0].legend(fontsize=9)
axes[0].grid(True)

axes[1].set_title('Varying Amplitude', fontsize=16)
axes[1].set_xlabel('X Values', fontsize=14)
axes[1].set_ylabel('Y Values', fontsize=14)
axes[1].legend(fontsize=9)
axes[1].grid(True)

axes[2].set_title('Varying Amplitude', fontsize=16)
axes[2].set_xlabel('X Values', fontsize=14)
axes[2].set_ylabel('Y Values', fontsize=14)
axes[2].legend(fontsize=9)
axes[2].grid(True)

plt.tight_layout()
plt.show()
```

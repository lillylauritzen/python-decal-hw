```python
import numpy as np
```

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)

#multiply by 4
arr_x4 = arr * 4
print(f"The list multiplied by 4 is {arr_x4}")

#mean
arr_mean = sum(arr)/len(arr)
print(f"The mean is {arr_mean}")

#median
sorted_list = sorted(list)
n = len(sorted_list)
mid1 = sorted_list[n//2-1]
mid2 = sorted_list[n//2]
arr_median = (mid1 + mid2) / 2
print(f"The median is {arr_median}")

#standard deviation
arr_mean = np.mean(arr)
std_pop = np.std(arr)
std_sample = np.std(arr, ddof=1)
print(f"The standard deviation for population is {std_pop}")
print(f"The standard deviation for sample is {std_sample}")

#arg min/arg max
arr_min = np.argmin(arr)
print(f"The index where the smallest number is occurs at index {arr_min}")

arr_max = np.argmax(arr)
print(f"The index where the largest number is occurs at index {arr_max}")

#corresponding values
min_idx = arr_min
print(f"The value at min index {min_idx} is {arr[min_idx]}")

max_idx = arr_max
print(f"The value at min index {max_idx} is {arr[max_idx]}")

#np.min() and np.max()
print(f"The value from np.min() is {np.min(arr)}")
print(f"The value from np.max() is {np.max(arr)}")
print("These numbers match from the previous step!")

#new array with indexing
condition = arr > 20
filtered_arr = arr[condition]
print(f"Values greater than 20: {filtered_arr}")
```

```python
matrix_list = [
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
]

matrix = np.array(matrix_list)
print(f"The shape of the array is {matrix.shape}")

print(f"The number of rows is {matrix.shape[0]} and the number of columns is {matrix.shape[1]}")

print(f"The 3rd column is {matrix[:, 2]}/n") 

print(f"The second row is {matrix[ 1, :]}/n")

matrix_doubled = matrix * 2
print("Matrix with elements multiplied by 2:")
print(matrix_doubled, "\n")

col_means = np.mean(matrix, axis = 0)
print(f"The mean of each column is {col_means}\n")

overall_mean = np.mean(matrix)
print(f"The overall mean of the array is {overall_mean}\n")

max_in_rows = np.max(matrix, axis=1)
min_in_cols = np.min(matrix, axis=0)
print(f"The max value in each row is {max_in_rows}")
print(f"The min value in each column is {min_in_cols}")

flattened = matrix.flatten()
print(f"The flattened array is {flattened}")
print(f"The new shape is {flattened.shape}\n")

flat_min_idx = np.argmin(flattened)
flat_max_idx = np.argmax(flattened)
print(f"The flat index of the smallest value is {flat_min_idx}")
print(f"the flat index of the largerst value is {flat_max_idx}")

row_min, col_min = np.unravel_index(flat_min_idx, matrix.shape)
row_max, col_max = np.unravel_index(flat_max_idx, matrix.shape)
print(f"The smallest value coordinate is (row {row_min}, column {col_min})")
print(f"The largest value coordinate is (row {row_max}, column {col_max})\n")

greater_than_15 = matrix[matrix > 15]
print(f"The values that are greater than 15 are {greater_than_15}")

sorted_matrix = np.sort(matrix, axis = 1)
print("Matrix with each row sorted:")
print(sorted_matrix)
```

```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def check_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False 
    return True 

def get_prime_magnitude_rows(arr):
    vectorized_prime_check = np.vectorize(check_prime)
    prime_mask = vectorized_prime_check(arr)
    row_condition = np.any(prime_mask, axis =1)
    return arr[row_condition]
matching_rows = get_prime_magnitude_rows(arr)
print("These are the rows with at least one prime magnitude:")
print(matching_rows)

```

```python
def create_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[1::2, 1::2] = 1
    return board 
print(create_checkerboard())
```

```python
def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[0::2, 1::2] = 1
    board[1::2, 0::2] = 1
    return board
print(reverse_checkerboard())
```

```python
def expansion(arr, num_spaces):
    def expand_strings(s):
        spaces = " " * num_spaces
        return spaces.join([char for char in s])
    vectorized_expansion = np.vectorize(expand_strings)
    return vectorized_expansion(arr)


universe = np.array(['galaxy', 'clusters'])

print("Expansion with 1 space:")
print(expansion(universe, 1))

print("\nExpansion with 2 spaces:")
print(expansion(universe, 2))
```

```python
def second_brightest(arr):
    sorted_stars = np.sort(arr, axis =1)
    return sorted_stars[:, -2]

np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
print("Original array:")
print(stars)

print("\nSecond brightest star per cluster:")
print(second_brightest(stars))
```
